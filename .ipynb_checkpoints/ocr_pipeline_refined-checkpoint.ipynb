# General form OCR pipeline

This notebook reads images from `figs/`, saves cleaned images in
`preprocessed/`, and writes OCR results to `output/<image_name>/`.

Set `IMAGE_SOURCE` to `"original"` or `"processed"` to control which
image is sent to the OCR model. Conservative line removal is the general
default because it keeps printed field labels for structured extraction.
Blank-template subtraction is optional when only newly added marks are
required.

## 4. Preprocessing functions

In `lines` mode, long horizontal and vertical rules are removed while
printed labels remain available to identify fields and checkboxes.

In `template` mode, each completed form is aligned with the blank form
and the printed content is subtracted. This removes labels, borders, and
empty checkbox outlines while retaining newly added writing and ticks.


```python
def read_cv(path):
    image = cv2.imread(str(path), cv2.IMREAD_COLOR)
    if image is None:
        raise FileNotFoundError(f"Could not read image: {path}")
    return image


def align_to_template(image, template):
    image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    template_gray = cv2.cvtColor(template, cv2.COLOR_BGR2GRAY)

    orb = cv2.ORB_create(nfeatures=6000)
    points_a, features_a = orb.detectAndCompute(image_gray, None)
    points_b, features_b = orb.detectAndCompute(template_gray, None)
    if features_a is None or features_b is None:
        raise RuntimeError("Not enough features for form alignment.")

    matcher = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
    matches = sorted(matcher.match(features_a, features_b), key=lambda item: item.distance)
    matches = matches[:max(40, int(len(matches) * 0.25))]
    if len(matches) < 12:
        raise RuntimeError("Not enough reliable matches for form alignment.")

    source = np.float32([points_a[item.queryIdx].pt for item in matches]).reshape(-1, 1, 2)
    target = np.float32([points_b[item.trainIdx].pt for item in matches]).reshape(-1, 1, 2)
    matrix, _ = cv2.findHomography(source, target, cv2.RANSAC, 4.0)
    if matrix is None:
        raise RuntimeError("Form alignment failed.")

    height, width = template_gray.shape
    return cv2.warpPerspective(
        image, matrix, (width, height),
        flags=cv2.INTER_LINEAR,
        borderMode=cv2.BORDER_CONSTANT,
        borderValue=(255, 255, 255),
    )


def remove_small_objects(mask, min_area=4):
    count, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    clean = np.zeros_like(mask)
    for label in range(1, count):
        x, y, width, height, area = stats[label]
        if area >= min_area and width >= 2 and height >= 2:
            clean[labels == label] = 255
    return clean


def subtract_template(image, template):
    aligned = align_to_template(image, template)
    image_gray = cv2.cvtColor(aligned, cv2.COLOR_BGR2GRAY)
    template_gray = cv2.cvtColor(template, cv2.COLOR_BGR2GRAY)

    image_bg = cv2.GaussianBlur(image_gray, (0, 0), 15)
    template_bg = cv2.GaussianBlur(template_gray, (0, 0), 15)
    image_gray = cv2.divide(image_gray, image_bg, scale=255)
    template_gray = cv2.divide(template_gray, template_bg, scale=255)

    difference = cv2.subtract(template_gray, image_gray)
    threshold, _ = cv2.threshold(
        difference, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )
    threshold = int(np.clip(threshold, 12, 45))
    _, mask = cv2.threshold(difference, threshold, 255, cv2.THRESH_BINARY)

    edges = cv2.Canny(template_gray, 60, 160)
    edge_area = cv2.dilate(
        edges, cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))
    )
    mask[(edge_area > 0) & (difference < 50)] = 0
    mask = remove_small_objects(mask)
    return cv2.bitwise_not(mask)


def remove_lines(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    binary = cv2.adaptiveThreshold(
        gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY_INV, 31, 15
    )
    height, width = binary.shape
    horizontal_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT, (max(30, width // 25), 1)
    )
    vertical_kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT, (1, max(30, height // 35))
    )
    horizontal = cv2.morphologyEx(binary, cv2.MORPH_OPEN, horizontal_kernel)
    vertical = cv2.morphologyEx(binary, cv2.MORPH_OPEN, vertical_kernel)
    lines = cv2.bitwise_or(horizontal, vertical)
    lines = cv2.dilate(lines, np.ones((2, 2), np.uint8), iterations=1)
    text = cv2.bitwise_and(binary, cv2.bitwise_not(lines))
    text = cv2.morphologyEx(text, cv2.MORPH_CLOSE, np.ones((2, 2), np.uint8))
    return cv2.bitwise_not(text)


def preprocess_image(path, mode=PREPROCESS_MODE):
    image = read_cv(path)
    if mode == "template":
        if not TEMPLATE_PATH.is_file():
            raise FileNotFoundError(f"Blank template not found: {TEMPLATE_PATH}")
        processed = subtract_template(image, read_cv(TEMPLATE_PATH))
    else:
        processed = remove_lines(image)

    destination = PREPROCESSED_DIR / f"{path.stem}.png"
    if not cv2.imwrite(str(destination), processed):
        raise RuntimeError(f"Could not save processed image: {destination}")
    return destination
```

```python
import torch
from transformers import (
    AutoModelForMultimodalLM,
    AutoProcessor,
    BitsAndBytesConfig,
)


def load_model(model_id=MODEL_ID, use_4bit=False):
    if not torch.cuda.is_available():
        raise RuntimeError(
            "CUDA is unavailable to PyTorch. Fix CUDA before loading the model."
        )

    compute_dtype = (
        torch.bfloat16
        if torch.cuda.is_bf16_supported()
        else torch.float16
    )

    model_options = {
        "device_map": "auto",
        "low_cpu_mem_usage": True,
        "dtype": compute_dtype,
    }

    if use_4bit:
        model_options["quantization_config"] = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=compute_dtype,
            bnb_4bit_use_double_quant=True,
        )

    processor = AutoProcessor.from_pretrained(model_id)

    model = AutoModelForMultimodalLM.from_pretrained(
        model_id,
        **model_options,
    ).eval()

    return processor, model


processor, model = load_model(
    model_id=MODEL_ID,
    use_4bit=True,
)

print("Loaded:", MODEL_ID)
print("Model device:", model.device)
print(
    "Allocated GPU memory:",
    f"{torch.cuda.memory_allocated() / 1024**3:.2f} GB",
)
```

## 8. OCR generation and JSON parsing

The model receives three focused requests: identity details, residential
address, and correspondence address. Smaller tasks reduce field leakage
and confusion between the two address sections.


```python
RULES = '''Read only entered handwriting, typed entries, and visibly selected
checkboxes. Printed labels and instructions are never values. Assign fields by
physical location, not language. The Chinese-name field may contain Latin
letters. Never copy between name fields or address sections. Use state=filled
only for a visible entry, state=blank for a visibly empty field, and
state=uncertain otherwise. For blank or uncertain fields use an empty value.
Preserve Traditional Chinese and leading zeros. Form Contain borders which is given 
in 'figs/td320_form_borders_only.png'. When you are doing the extraction, 
you should ignore the borders and only focus on the content inside the form.
You must not add any extra fields or change the field names.
You should extract borders and content inside the form to make sure
the extracted content is complete and accurate.
English and Chinse names only contain latters.
The orginal form is 'td320_e-fillable_eng.pdf', you may refer to it for 
the field locations and borders.
Name fields only contain letters.
Name fields can not contain numbers, symbols, or punctuation.
Be careful about differences between '1', 'I', 'i' with '|' used in form to 
separate the fields, you can refer to the original form for guidance whenever you are 
uncertain.
Return only valid JSON.'''

PROMPTS = {
    "identity": RULES + '''
Extract these fields:
{"name_english":{"value":"","state":"filled|blank|uncertain"},
 "name_chinese":{"value":"","state":"filled|blank|uncertain"},
 "identity_document_number":{"value":"","state":"filled|blank|uncertain"},
 "telephone":{"value":"","state":"filled|blank|uncertain"},
 "agent_name":{"value":"","state":"filled|blank|uncertain"},
 "agent_identity_document_number":{"value":"","state":"filled|blank|uncertain"},
 "date":{"value":"","state":"filled|blank|uncertain"}}''',
    "residential": RULES + '''
Extract only Residential Address. Region must be Hong Kong, Kowloon,
New Territories, or empty:
{"flat":{"value":"","state":"filled|blank|uncertain"},
 "floor":{"value":"","state":"filled|blank|uncertain"},
 "block":{"value":"","state":"filled|blank|uncertain"},
 "building":{"value":"","state":"filled|blank|uncertain"},
 "street":{"value":"","state":"filled|blank|uncertain"},
 "district":{"value":"","state":"filled|blank|uncertain"},
 "region":{"value":"","state":"filled|blank|uncertain"}}''',
    "correspondence": RULES + '''
Extract only Correspondence Address. Do not repeat Residential Address.
Region must be Hong Kong, Kowloon, New Territories, or empty:
{"flat":{"value":"","state":"filled|blank|uncertain"},
 "floor":{"value":"","state":"filled|blank|uncertain"},
 "block":{"value":"","state":"filled|blank|uncertain"},
 "building":{"value":"","state":"filled|blank|uncertain"},
 "street":{"value":"","state":"filled|blank|uncertain"},
 "district":{"value":"","state":"filled|blank|uncertain"},
 "region":{"value":"","state":"filled|blank|uncertain"}}''',
}


def prepare_pil(path):
    image = ImageOps.exif_transpose(Image.open(path)).convert("RGB")
    if max(image.size) > MAX_IMAGE_EDGE:
        scale = MAX_IMAGE_EDGE / max(image.size)
        image = image.resize(
            (round(image.width * scale), round(image.height * scale)),
            Image.Resampling.LANCZOS,
        )
    return image


def generate_text(image, prompt):
    messages = [{"role": "user", "content": [
        {"type": "image", "image": image},
        {"type": "text", "text": prompt},
    ]}]
    inputs = processor.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_dict=True,
        return_tensors="pt",
    ).to(model.device)

    with torch.inference_mode():
        output = model.generate(
            **inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.05,
        )

    tokens = output[0, inputs["input_ids"].shape[1]:]
    if len(tokens) >= MAX_NEW_TOKENS:
        raise RuntimeError("Model response reached the token limit.")
    return processor.decode(tokens, skip_special_tokens=True).strip()


def parse_json(text):
    decoder = json.JSONDecoder()
    for index, character in enumerate(text):
        if character == "{":
            try:
                value, _ = decoder.raw_decode(text[index:])
                if isinstance(value, dict):
                    return value
            except json.JSONDecodeError:
                pass
    raise ValueError("No valid JSON object found in the model response.")
```

## 9. Validate and combine model responses

This stage rejects printed labels, normalizes phone/date formatting,
validates regions, and records any name fallback. It never invents a
missing district or address.


```python
VALID_REGIONS = {"hong kong": "Hong Kong", "kowloon": "Kowloon",
                 "new territories": "New Territories"}
PRINTED_LABELS = {
    "name", "identitydocumentno", "flatroom", "floor", "blocktower",
    "nameofbuildingestate", "numberandnameofstreetorvillage", "district",
    "daytimecontacttelno", "date",
}


def clean_leaf(node, field, review):
    if not isinstance(node, dict):
        review.append({"field": field, "issue": "Invalid response structure"})
        return {"value": "[UNCERTAIN]", "state": "uncertain"}

    value = unicodedata.normalize("NFKC", str(node.get("value", ""))).strip()
    state = str(node.get("state", "uncertain")).strip().lower()
    if state not in {"filled", "blank", "uncertain"}:
        state = "uncertain"
    if state != "filled":
        return {"value": "" if state == "blank" else "[UNCERTAIN]", "state": state}
    if not value:
        review.append({"field": field, "issue": "Filled field has no value"})
        return {"value": "[UNCERTAIN]", "state": "uncertain"}

    label = re.sub(r"[^a-z]", "", value.lower())
    if label in PRINTED_LABELS:
        review.append({"field": field, "issue": "Printed label returned as value"})
        return {"value": "[UNCERTAIN]", "state": "uncertain"}
    return {"value": value, "state": "filled"}


def combine_results(results):
    review = []
    identity_fields = (
        "name_english", "name_chinese", "identity_document_number",
        "telephone", "agent_name", "agent_identity_document_number", "date",
    )
    address_fields = ("flat", "floor", "block", "building", "street", "district", "region")

    fields = {
        key: clean_leaf(results.get("identity", {}).get(key), key, review)
        for key in identity_fields
    }
    for section in ("residential", "correspondence"):
        fields[f"{section}_address"] = {
            key: clean_leaf(
                results.get(section, {}).get(key),
                f"{section}_address.{key}",
                review,
            ) for key in address_fields
        }

    def is_filled(item):
        return item["state"] == "filled" and bool(item["value"])

    # Copy an existing name only when the other name location is blank.
    for target, source in (("name_english", "name_chinese"),
                           ("name_chinese", "name_english")):
        if not is_filled(fields[target]) and is_filled(fields[source]):
            fields[target] = {"value": fields[source]["value"], "state": "fallback"}
            review.append({"field": target, "issue": f"Copied from {source}"})

    phone = fields["telephone"]
    if is_filled(phone):
        phone["value"] = re.sub(r"[\s().-]", "", phone["value"])
        if not re.fullmatch(r"\d{8}", phone["value"]):
            review.append({"field": "telephone", "issue": "Expected 8 digits"})

    date = fields["date"]
    if is_filled(date):
        date["value"] = re.sub(r"\s*([/.-])\s*", r"\1", date["value"])
        valid = False
        for date_format in ("%d/%m/%Y", "%d-%m-%Y", "%d.%m.%Y"):
            try:
                datetime.strptime(date["value"], date_format)
                valid = True
                break
            except ValueError:
                pass
        if not valid:
            review.append({"field": "date", "issue": "Invalid date"})

    for section in ("residential_address", "correspondence_address"):
        region = fields[section]["region"]
        if is_filled(region):
            canonical = VALID_REGIONS.get(region["value"].lower())
            if canonical:
                region["value"] = canonical
            else:
                region.update(value="[UNCERTAIN]", state="uncertain")
                review.append({"field": f"{section}.region", "issue": "Invalid region"})

    def values_only(item):
        if isinstance(item, dict) and set(item) == {"value", "state"}:
            return item["value"]
        if isinstance(item, dict):
            return {key: values_only(value) for key, value in item.items()}
        return item

    def states_only(item):
        if isinstance(item, dict) and set(item) == {"value", "state"}:
            return item["state"]
        if isinstance(item, dict):
            return {key: states_only(value) for key, value in item.items()}
        return item

    audit = {
        "status": "needs_review" if review else "passed_automated_checks",
        "review_items": review,
        "field_states": states_only(fields),
    }
    return values_only(fields), audit
```

## 10. Write English and Chinese output files

Values remain unchanged between languages; only field labels are
translated. This preserves a Chinese-name field containing Latin letters.
Raw model responses and an audit file are also retained for review.


```python
CHINESE_LABELS = {
    "name_english": "英文姓名", "name_chinese": "中文姓名",
    "identity_document_number": "身份證明文件號碼",
    "residential_address": "住址", "correspondence_address": "通訊地址",
    "telephone": "日間聯絡電話", "agent_name": "代理人姓名",
    "agent_identity_document_number": "代理人身份證明文件號碼",
    "date": "日期", "flat": "室", "floor": "樓", "block": "座",
    "building": "大廈或屋苑名稱", "street": "門牌號數及街道或鄉村名稱",
    "district": "地區", "region": "區域",
}
ENGLISH_LABELS = {
    "name_english": "English name", "name_chinese": "Chinese-name field",
    "identity_document_number": "Identity document number",
    "residential_address": "Residential address",
    "correspondence_address": "Correspondence address",
    "telephone": "Telephone", "agent_name": "Agent name",
    "agent_identity_document_number": "Agent identity document number",
    "date": "Date", "flat": "Flat/Room", "floor": "Floor",
    "block": "Block/Tower", "building": "Building/Estate",
    "street": "Street/Village", "district": "District", "region": "Region",
}


def translate_keys(value):
    if isinstance(value, dict):
        return {CHINESE_LABELS.get(key, key): translate_keys(item)
                for key, item in value.items()}
    return value


def render_text(data, labels):
    lines = []
    for key, value in data.items():
        label = labels.get(key, key)
        if isinstance(value, dict):
            lines.append(f"{label}:")
            lines.extend(f"  {labels.get(name, name)}: {entry}"
                         for name, entry in value.items())
        else:
            lines.append(f"{label}: {value}")
    return "\n".join(lines) + "\n"


def write_outputs(image_path, data, audit, raw):
    folder = OUTPUT_DIR / image_path.stem
    folder.mkdir(parents=True, exist_ok=True)
    name = image_path.stem

    files = {
        f"{name}_English.txt": render_text(data, ENGLISH_LABELS),
        f"{name}_Chinese.txt": render_text(data, CHINESE_LABELS),
        f"{name}_English.json": json.dumps(data, ensure_ascii=False, indent=2) + "\n",
        f"{name}_Chinese.json": json.dumps(translate_keys(data), ensure_ascii=False, indent=2) + "\n",
        f"{name}_audit.json": json.dumps(audit, ensure_ascii=False, indent=2) + "\n",
        f"{name}_raw.json": json.dumps(raw, ensure_ascii=False, indent=2) + "\n",
    }
    for filename, content in files.items():
        (folder / filename).write_text(content, encoding="utf-8")

    # Confirm that both final JSON files can be reopened.
    for language in ("English", "Chinese"):
        json.loads((folder / f"{name}_{language}.json").read_text(encoding="utf-8"))
    return folder
```

## 11. Process one image or the complete batch

`process_image` performs OCR and writes one output folder. `run_batch`
continues after individual failures and records the overall status in
`output/batch_summary.json`.


```python
def process_image(original_path):
    started = time.perf_counter()
    ocr_path = choose_ocr_path(original_path)
    image = prepare_pil(ocr_path)
    raw = {}
    parsed = {}

    try:
        for name, prompt in PROMPTS.items():
            raw[name] = generate_text(image, prompt)
            parsed[name] = parse_json(raw[name])

        data, audit = combine_results(parsed)
        audit.update({
            "original_image": str(original_path),
            "ocr_image": str(ocr_path),
            "image_source": IMAGE_SOURCE,
            "model": MODEL_ID,
            "elapsed_seconds": round(time.perf_counter() - started, 2),
        })
        folder = write_outputs(original_path, data, audit, raw)
        return {"image": original_path.name, "status": audit["status"],
                "output": str(folder)}
    except Exception as error:
        folder = OUTPUT_DIR / original_path.stem
        folder.mkdir(parents=True, exist_ok=True)
        failure = {
            "status": "error", "image": str(original_path),
            "ocr_image": str(ocr_path), "error": str(error), "raw": raw,
        }
        (folder / f"{original_path.stem}_error.json").write_text(
            json.dumps(failure, ensure_ascii=False, indent=2) + "\n",
            encoding="utf-8",
        )
        raise


def run_batch(image_paths):
    summary = []
    for index, path in enumerate(image_paths, start=1):
        print(f"[{index}/{len(image_paths)}] {path.name}")
        try:
            result = process_image(path)
        except Exception as error:
            result = {"image": path.name, "status": "error", "error": str(error)}
        summary.append(result)
        print(" ", result["status"])

    summary_path = OUTPUT_DIR / "batch_summary.json"
    summary_path.write_text(
        json.dumps(summary, ensure_ascii=False, indent=2) + "\n",
        encoding="utf-8",
    )
    return summary
```

## 12. Test one example

Run `Chan` first to verify preprocessing, inference, parsing, and output
creation before launching the complete batch.


```python
example_name = "Chan"
example_path = next(
    (path for path in images if path.stem.lower() == example_name.lower()),
    None,
)
if example_path is None:
    raise FileNotFoundError(f"No image named {example_name} found in {FIGS_DIR}.")

example_result = process_image(example_path)
print(json.dumps(example_result, ensure_ascii=False, indent=2))
```

## 13. Process the remaining images

After reviewing the example output, run this cell. `Chan` is not sent to
the model a second time. The final summary includes both the example and
the remaining images.


```python
remaining_images = [path for path in images if path != example_path]
batch_summary = [example_result]
if remaining_images:
    batch_summary.extend(run_batch(remaining_images))

(OUTPUT_DIR / "batch_summary.json").write_text(
    json.dumps(batch_summary, ensure_ascii=False, indent=2) + "\n",
    encoding="utf-8",
)
print(json.dumps(batch_summary, ensure_ascii=False, indent=2))
```

## Output interpretation

- `passed_automated_checks` means the schema and formatting checks passed;
  it does not guarantee that every character is correct.
- `needs_review` means the audit file lists uncertain, copied, or invalid
  fields requiring comparison with the original image.
- Checkbox-derived regions should always be visually verified.
